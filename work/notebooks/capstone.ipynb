## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

*Dataset: FlyRank pseudonymized warehouse release (v20260703).
Tables Used: dim_content (for page metadata) and fact_content_daily_performance (for time-series performance metrics).
Date Windows: Features were engineered from a mid-panel month (March 2026), leaving the final month (June 2026) untouched for holdout testing.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

*Task: Supervised Binary Classification scoring pages to rank them.
Target Definition: is_declining (1 if future impressions drop by >20% compared to the historical window, 0 otherwise).
Features: content_age_days, impressions_90d, ctr, and avg_position.
Validation Design: To prevent the model from memorizing client-specific baseline traffic, we utilized GroupShuffleSplit on client_id.
Leakage Checks: Confirmed no target-derived variables or product scores exist in the feature matrix.*

```python
print("Methodology: Random Forest Classifier validated via GroupShuffleSplit to ensure honest performance on unseen clients.")
```

## 5. Limitations

*What this work cannot claim.*

*What this work cannot claim:

1-Causality: This is an observational study. A high risk score does not mathematically prove that a rewrite will recover the traffic.  

2-Completeness: The model is blind to off-page SEO factors, such as backlink decay or broad Google core updates.  

3-Algorithm Decoding: We are measuring directional risk for decision-support, not reverse-engineering Google's ranking factors.*

```python
print("Limitation Note: The model acts as a highly educated compass, not a crystal ball.")
```
