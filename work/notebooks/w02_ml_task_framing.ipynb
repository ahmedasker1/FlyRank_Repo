## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
#The primary success metric is Precision@K (e.g., Precision@50). Accuracy is misleading here; "good" means that out of the top 50 pages we recommend the team to review, the vast majority are genuinely declining
# We only care about the top K results matching human capacity.
K = 50
print(f"Metric: Precision@{K} - Out of the top {K} ranked pages, how many are true positives?")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
#ML beats a fixed rule because decay patterns are multi-dimensional. An if-statement (e.g., age > 180 AND impressions > 500) sets rigid thresholds, missing pages that might be slightly newer but suffering a severe CTR drop. ML weighs these interacting signals dynamically.
# A fixed rule fails to capture nuanced interactions between features.
print("ML captures interactions (e.g., age * CTR * volume) that a rigid IF-statement misses.")
```
